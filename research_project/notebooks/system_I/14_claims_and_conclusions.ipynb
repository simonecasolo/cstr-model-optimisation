## 1. CSTR model and scenario overview

The study is built on a 3-state closed-loop CSTR (propylene-oxide hydrolysis, Fogler M13)
with a PI temperature controller (Tsp = 312.5 K, Kp = 150, τi = 10). The inference task
is reduced to two physically meaningful scalars:

- **α ∈ [0.4, 1.2]** — fractional catalyst activity (acid neutralisation).
- **β ∈ [0.4, 1.2]** — fractional jacket UA (Kern–Seaton fouling).

The physically meaningful range is [0.4, 1.0]; the prior upper bound is extended to
1.2 as a numerical buffer for the neural density estimator so that the nominal healthy
state (α=β=1.0) sits in the interior of the prior support. Posterior samples above 1.0
may occur due to numerical effects but are not physically meaningful.

UA and k0 are fixed clean-service design constants (notebook 05 §5 showed they are
structurally non-identifiable from CL data because they only enter the ODE through
the products α·k0 and β·UA).

**Noise model.** The simulator adds two layers of noise (see `simulator.py`):

- *Process noise* — continuous-time SDE diffusion applied via Euler-Maruyama:
  σ_C = 0.0005 mol/L/√min, σ_T = σ_Tc = 0.1 K/√min. These are scaled down
  from the literal spec values (0.1 mol/L/min, 10 K/min) which refer to a
  discrete-time fixed-step interpretation in legacy code; as proper SDE
  diffusion coefficients they produce a steady-state OU standard deviation
  of ~0.5 K on T — visible but well below the fault signal magnitude.
- *Sensor noise* — Gaussian, 0.5% of each channel's running maximum,
  applied post-simulation (spec §3.5).

**Sub-claims supported by this block:**

1. *The eight scenarios span the operating envelope the paper claims to cover* —
   healthy (Sc0/Sc1), single-fault (Sc2 fouling, Sc3 decay), combined (Sc4),
   controller-saturated (Sc5), open-loop mismatch (Sc6), sensor drift (Sc7) and a
   30-day continuous degradation stream (Sc8).
2. *The simulator is faithful enough to be the data-generating process for SBI* —
   steady-state values for Sc0/Sc1 match the Fogler reference (C ≈ 0.018 mol/L,
   T ≈ 312 K, Tc ≈ 299 K).

**Evidence in the figures below:**

- **Fig 1a** plots the four-channel windows [C, T, Tc, Qc] for every scenario.
  The healthy/closed-loop traces (Sc1) sit on Tsp with bounded Qc oscillations,
  Sc2 shows the *fouling signature* (T held at Tsp, Qc walks up to compensate),
  Sc3 shows the *decay signature* (Qc drops because less heat is generated),
  and Sc5 shows the saturation envelope (Qc clipped at Qc_max).  This visually
  confirms each scenario produces the dynamics specified in the research spec.
- **Fig 1b** is the headline summary of the closed-loop compensation mechanism:
  reactor temperature deviation from Tsp (left panel) and mean coolant flow Qc
  (right panel) across all scenarios. The left panel shows that the PI controller
  keeps T within ±2 K of setpoint for most scenarios — the fault signal is
  *masked* in the temperature channel. The right panel shows that Qc carries
  the fault signature instead: fouling scenarios (Sc2, Sc5) drive Qc up to
  compensate for reduced heat transfer, while decay (Sc3) lowers it. Sc4
  (combined) pushes Qc near the saturation limit (Qc_max = 400). This
  motivates why closed-loop SBI must exploit the control channel, not just
  the process states.

## 2. Summary statistics and feature space (nb03)

A 29-D summary vector compresses each 60-min window of [C, T, Tc, Qc] into a single
embedding for the neural posterior. The vector decomposes as:
5 base stats × 4 channels (20) + final-window means (4) + control aggregates (3) + the
two physics-informed proxies `UA_eff_proxy = (T_mean−Tc_mean)/Qc_mean ∝ 1/(β·UA)` and
`k0_eff_proxy = log(C_mean/(Ci−C_mean)) ∝ 1/(α·k0)`.

**Sub-claims:**

1. *The 29-D summary statistics carry sufficient identifying information for
   both fault classification and parameter recovery* — they are not a
   bottleneck for the inference task. A separate experiment (nb04b) trained an
   SBI posterior with a learned CNN embedding on raw (120×4) time series,
   bypassing the hand-crafted summaries entirely. The learned embedding
   performed no better on β (0.621 vs 0.616 for Sc2) and worse on α
   (e.g. Sc3: α_mean = 1.01 vs 0.70 true), confirming the 29-D vector is
   not losing information relative to an end-to-end learned alternative.
2. *The physics-informed pair carries the dominant share of identifying information
   for α and β* — this justifies their inclusion and underpins the
   ≥ 98% LDA scenario-classification result (nb03 §7b).

**Evidence:**

- **Fig 2a (PCA)** — projecting the 29-D vector onto the first two principal components
  already separates Sc1 (healthy), Sc2 (fouling), Sc3 (decay) and Sc5 (saturation) into
  distinct clusters. *Supports sub-claim 1.* The overlap between Sc4 (combined) and the
  edges of Sc2/Sc3 is the linear-projection signature of the combined-fault degeneracy
  that the full posterior resolves (see §6).
- **Fig 2b (t-SNE)** — non-linear embedding gives essentially perfect cluster
  separation for the single-fault scenarios; the residual mixing for Sc4/Sc7
  foreshadows the (α≈1, β=0.85) ambiguity in the moderate-severity regime.
- **Fig 2c (physics-feature scatter)** — `UA_eff_proxy` vs `k0_eff_proxy` alone
  separates Sc1/Sc2/Sc3/Sc5 into four well-resolved manifolds. *Directly supports
  sub-claim 2.* These two scalars contain almost all of the (α, β) identifying
  information — the rest of the 29-D vector mostly contributes robustness against
  noise, not new information.
- **Fig 2d (mutual information)** — ranking confirms `k0_eff_proxy` is the top MI
  feature for α and `UA_eff_proxy` for β, by a wide margin. The control aggregates
  (Qc saturation fractions, control integral) are the next tier, providing the
  closed-loop-specific signal that the open-loop SBI in §5 cannot exploit.

*Where the data is silent:* mutual information measures pairwise dependence, not
joint identifiability — Fig 2c/2d do **not** by themselves prove (α, β) are
jointly identifiable; that is the job of §4 (NUTS posteriors) and §6.

*Trade-off with learned embeddings:* while a CNN embedding could in principle
discover features that the hand-crafted set misses, the nb04b result shows it
does not improve recovery on this problem. The hand-crafted summaries have the
advantage of **explainability** — each feature has a direct physical
interpretation (e.g. `UA_eff_proxy ∝ 1/(β·UA)`), which is valuable for
diagnosing why the posterior behaves as it does (§4, §6).

### NPE as a Direct-Method Analogue

Classical closed-loop identification distinguishes three families of methods: *direct*
(operate on raw input-output data without requiring the controller model), *indirect*
(back-calculate the plant by estimating the closed-loop model and the known controller),
and *joint input-output* (model the entire closed-loop system jointly)
(Forssell & Ljung, 1999, section 2.3).

NPE as deployed here falls cleanly into the **direct method** class: the neural
posterior estimator is trained on measured $(C, T, T_c, Q_c)$ histories --- without
utilising the PI controller's internal equations (gains, integral state, setpoint) ---
and approximates $p(\alpha, \beta \mid x)$ end-to-end. This classification is not
merely taxonomic. It explains two key properties of the approach:

1. **Universal applicability to arbitrary feedback.** The direct method works for any
   controller topology --- linear, nonlinear, saturating --- because it does not need to
   model the controller. The CSTR's clipped PI output and the resulting input-noise
   coupling (section 6) affect the posterior quality but do not invalidate the inference
   procedure.
2. **UPDATED (2026-08-18, Item 8).** *Two distinct effects, previously conflated as
   one "inherited bias structure."* The classical direct method in closed loop is
   subject to bias from the non-zero input-noise cross-spectrum $\Phi_{eu}$
   (Forssell & Ljung, 1999, Eq. 3.44) --- but the large ($-0.08$ to $-0.15$) bias
   this notebook originally reported and attributed entirely to that mechanism was
   shown (matched-protocol retraining, Item 8) to be substantially a *separate*,
   avoidable training/evaluation distribution-shift bug (fixed warm-start training
   vs. scenario-specific evaluation data), not the $\Phi_{eu}$ effect itself. Once
   that bug is fixed, SBI's residual Sc2 bias is $-0.0023$; EKF and UKF, whose
   state initialisation was never subject to the bug, independently retain a
   genuine but much smaller $\approx-0.09$ bias --- this smaller, method-independent
   residual, not the original $-0.08$ to $-0.15$, is the credible remaining
   signature of closed-loop information reduction.

### Control-Theoretic Explanation: Why OL Training Fails ($\Phi_{ue}$ Mismatch)

The W1_beta degradation for OL-trained SBI on Sc2 has a precise
control-theoretic cause that is **distinct from the (now much smaller) residual
beta bias discussed in section 6**.

Under **closed-loop** operation, the PI controller computes:

$$Q_c(t) = Q_{c0} + K_p\bigl(T_{sp} - T_{meas}(t)\bigr) + \frac{K_p}{\tau_i}\int_0^t\bigl(T_{sp} - T_{meas}(s)\bigr)\,ds$$

where $T_{meas}(t) = T(t) + \varepsilon_T(t)$ includes sensor noise. This makes
$Q_c$ an explicit function of the measurement noise $\varepsilon_T$, creating a
non-zero **input-noise cross-spectrum**:

$$\Phi_{ue}(\omega) = -K(e^{i\omega})\,S_0(e^{i\omega})\,H_0(e^{i\omega})\,\lambda_0 \neq 0$$

Under **open-loop** operation, $Q_c$ is set independently of $T$, so
$\Phi_{ue}(\omega) = 0$ throughout.

| Training data | $\Phi_{ue}$ | Deployment data | Result (matched protocol) |
|---|---|---|---|
| Open-loop | 0 | Closed-loop | $\Phi_{ue} \neq 0$ -- **mismatch** -> beta_hat=0.152 (true 0.70) |
| Closed-loop | $\neq 0$ | Closed-loop | Matched -- beta_hat=0.698 (true 0.70) |
| Closed-loop | $\neq 0$ | Open-loop (Sc6) | $\Phi_{ue} = 0$ -- **mismatch** -> accuracy collapse (1.00 to 0.04) |

An SBI network trained on open-loop simulations learns summary statistics whose
joint distribution assumes $Q_c$ and $T$ noise are independent. When deployed on
closed-loop plant data, the PI-induced correlation between $Q_c$ transients and
$T$ noise has a different statistical structure than anything the network saw
during training. The posterior is therefore evaluated far from its training support,
producing the extreme bias (beta_hat ~= 0.15 vs truth = 0.70) seen in Fig 5a.

**This remains a distinct and avoidable failure mode.** CL training eliminates
the $\Phi_{ue}$ mismatch; matched-protocol CL-SBI now recovers beta with bias
$-0.0023$ (Sc2), i.e. this mode-mismatch mechanism is essentially the *whole*
CL-vs-OL story for the training-protocol-clean posterior -- **updated
(2026-08-18, Item 8): the "residual -0.08 beta bias which no training strategy can
remove" claimed below in the original version of this cell was itself
substantially the training/evaluation warm-start bug diagnosed in section 3, not
a separate irreducible Fisher-information floor.** The genuine, much smaller
residual information deficit is now best evidenced by EKF/UKF (section 6, section 9),
not by this CL/OL comparison.

  artefact, not a Fisher-information floor.** SBI, NUTS, and the CNN-embedding
  variant all generated their synthetic training/reference data from a single
  fixed healthy warm-start, while every evaluation trajectory started from its
  own scenario-specific steady state -- a train/test distribution mismatch. Fixing
  this (per-draw, scenario-specific warm start) collapsed SBI's Sc2 bias from
  $-0.1488$ to $-0.0023$, NUTS's from $\approx-0.10$ to $-0.0002$, and brought SBC
  from rejecting uniformity ($p=0.016/0.014$) to passing ($p=0.21/0.13$).
- **EKF and UKF, which were never subject to that bug** (their state initialises
  from the real observation, not a simulated training protocol), independently
  retain a genuine $\approx-0.09$ Sc2 beta bias under the matched protocol. This,
  not the original six-method table below, is the credible remaining evidence for
  *some* real closed-loop information deficit.
- **The Fisher-information ratio itself is not a fixed "250-500x."** Re-derived
  under the matched protocol (nb15), $\widetilde I_{\alpha\alpha}/\widetilde
  I_{\beta\beta} \approx 236\times$ at the healthy operating point but only
  $\approx 12\times$ at the Sc2 fouling condition used throughout the rest of the
  paper, and continuously between $\approx5\times$ and $\approx170\times$ across
  the beta prior range -- i.e. beta's own Fisher information rises sharply as the
  fault deepens, a real physical effect (larger, more separable coolant-flow
  response), not noise.
- **The ratio is also methodology-dependent** (nb33, this session): a
  diagonal-covariance FIM estimator gives $\approx0.9\times$ at Sc2 on both the
  29-D summary vector and the raw 480-D trajectory -- close to parity -- versus
  $\approx11$--$12\times$ from the full-covariance estimator on the same summary
  vector. A substantial share of the apparent asymmetry is attributable to
  cross-feature correlations that a diagonal approximation discards, not to the
  underlying plant dynamics alone.
- **The simplified 4-observable analytical model gives the *opposite* ordering**
  ($\approx0.6\times$ at Sc2, i.e. beta *more* identifiable than alpha under this
  reduction) -- not "$\approx600\times$, broadly consistent" as originally
  reported here. This is a genuine representation-sensitivity finding (the same

  omits the dynamic/transient features that make up 25 of the 29 summary

  $+14{,}859$ L/min vs. the directly simulated steady state).

**What survives unchanged:** the exact physical identity

### Control-Theoretic Perspective: Why the Data Is Information-Poor for Beta

**Note (2026-08-18): the "250-500x smaller" figure below is the retracted headline
number (see section 6 above) -- read this argument as establishing the qualitative
*direction* of the asymmetry (feedback suppresses low-frequency excitation of the
controlled variable), not its magnitude.** The frequency-domain argument itself
(Forssell & Ljung, 1999) is unaffected by the protocol correction; only the
numeric ratio it was originally paired with has changed.

Define the **sensitivity function** of the PI feedback loop:

$$S_0(q) = \frac{1}{1 + G_0(q)\,K(q)}$$

where $G_0(q)$ is the (linearised) plant transfer function from $Q_c$ to $T$, and
$K(q)$ is the PI controller. The noise-free portion of the input spectrum --- the
part of $Q_c$ variation that carries independent information about the plant ---
satisfies:

$$\Phi_u^r(\omega) = |S_0(e^{i\omega})|^2 \, \Phi_r(\omega)$$

where $\Phi_r(\omega)$ is the power spectrum of the reference (setpoint) signal.
The asymptotic estimation variance scales as $\text{Var}(\hat\theta) \propto
\Phi_v(\omega) / \Phi_u^r(\omega)$.

Two compounding effects drive $\Phi_u^r \to 0$ for the CSTR at fixed setpoint:

1. **No reference excitation**: the setpoint is constant, so $\Phi_r(\omega) = 0$
   at all frequencies where beta would need to be excited.
2. **Integral action**: the PI integral action forces $S_0(e^{i\omega}) \to 0$ at
   low frequencies (the frequencies where steady-state heat-balance information
   lives).

Both factors multiply in $\Phi_u^r = |S_0|^2 \Phi_r \to 0$, driving down (not, per
the corrected finding above, all the way to zero) the asymptotic information for
beta relative to alpha.

This is the **frequency-domain dual of the Jacobian collapse** established in
section 6: $\partial T_{ss}/\partial\beta = 0$ in steady-state corresponds to
$|S_0|^2\Phi_r \to 0$ in frequency domain -- both exact, protocol-independent
statements. The specific *magnitude* this predicts is where the operating-point-
and methodology-dependence documented in section 6 above enters: the frequency-
domain argument explains the *sign and existence* of the asymmetry, not a single
universal ratio.

### Contribution framing

**UPDATED (2026-08-18, Item 8): the quote block below is RETRACTED and replaced
--- it stated the original, since-substantially-retracted "250-500x, irreducible,
four-method-confirmed" claim as the paper's headline. The corrected framing
(matching `main.tex`'s current abstract) is:**

> We show that an apparent closed-loop Fisher-information deficit for the fouling
> parameter, initially observed to agree across four independent inference
> paradigms (SBI, NUTS, EKF, UKF), is substantially a shared training-protocol
> artefact for two of those methods (SBI, NUTS) rather than a genuine,
> method-independent limitation; a smaller, methodology- and operating-point-
> dependent closed-loop information deficit remains once the protocol is
> corrected, evidenced most credibly by EKF/UKF (never subject to the bug).
> Despite this smaller residual limitation, amortised SBI delivers real-time
> probabilistic fault classification (closed-loop macro-F1 = 0.927) at
> orders-of-magnitude lower per-window latency than MCMC, and now with *better*
> point-accuracy than classical Kalman-filter baselines, not just comparable
> accuracy at a variance cost.

The paper's contribution is **not** the discovery of a closed-loop identifiability
limitation --- that principle has been established since Gustavsson, Ljung &
Soderstrom (1977) and is well-known in system identification. The contribution is
the careful *disentangling* of a real, smaller closed-loop information effect from
a much larger, avoidable training-protocol artefact that had been conflated with
it -- itself a methodologically useful cautionary result for SBI practice generally
(matched training/evaluation protocols matter, and simulation-based calibration,
correctly interpreted, can catch this class of bug).
